#Neural Network Algorithm - Activation Functions


```python
#import necessary libraries
import numpy as np
import matplotlib.pyplot as plt


z = np.linspace(-10, 10, 100)  # generates 100 evenly spaced values from -10 to 10

#using Activation Functions
sigmoid = 1 / (1 + np.exp(-z))
relu = np.maximum(0, z)

#ploting graph between the two functions
plt.plot(z, sigmoid, label="Sigmoid")
plt.plot(z, relu, label="ReLU")
plt.xlabel("z")
plt.ylabel("Output")
plt.title("Sigmoid vs ReLU")
plt.legend()
plt.axhline(0, color="gray", linewidth=0.5)  # reference line at y=0
plt.axvline(0, color="gray", linewidth=0.5)  # reference line at x=0
plt.show()
```

```python
#caculating and plotting softmax output

z_values = np.array([-2.87382211 , 1.41051157 , -0.33235825])  #these are the 3 raw output neuron z-values

exp_z = np.exp(z_values)
softmax_output = exp_z / np.sum(exp_z)

print(softmax_output)
print(softmax_output.sum())  # should be exactly 1.0 (100%)

plt.bar(["setosa", "versicolor", "virginica"], softmax_output)
plt.ylabel("Probability")
plt.title("Softmax Output")
plt.show()
```
